```python
import pandas as pd

# Carico i dati
competitions = pd.read_json(r"data\raw\competitions.json")

#per visualizzare meglio il mask
pd.set_option('display.width', 1000)
pd.set_option('display.expand_frame_repr', False)


#print(competitions['competition_id'].value_counts())
#print(competitions.groupby("competition_name")["season_name"].unique())

"""
Per cercare gli ID che mi servono in base al nome:
print(competitions[competitions["competition_name"].str.contains("World|Euro", case=False, na=False)]
      [["competition_id","season_id","competition_name","season_name"]]             #non mi visualizza tutte le colonne, solo quelle riportate
      .drop_duplicates()                                                            #garantisce che ogni riga sia unica
      .sort_values(["competition_name","season_name"]))
"""

#meglio filtrare per coppie (competition_id,season_id)
coppie_target=[
    (43,3),        #FIFA World Cup 2018
    (43,106),      #FIFA World Cup 2022
    (55,43),       #UEFA Euro 2020
    (55,282)       #UEFA Euro 2024
]
"""
Creo una Serie Booleana:
    -seleziono le colonne competition_id e season_id
    -trasformo ogni riga in una tupla con .apply e axis=1 (competition_id,season_id)
    -verifico se fa match con .isin
"""
mask = competitions[['competition_id','season_id']].apply(tuple,axis=1).isin(coppie_target)


#applico il filtro

df_targets = competitions.loc[mask].copy()

#debug controllo che siano 4 righe e che siano giuste
print(len(df_targets))
print(df_targets[["competition_id","season_id","competition_name","season_name"]])
```
